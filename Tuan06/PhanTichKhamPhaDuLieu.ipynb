### 1.1.1. Ôn tập lý thuyết
#### + Thống kê mô tả là gì? Nó khác gì với thống kê suy luận (inferential statistics)?
Thống kê mô tả (Descriptive Statistics) là phương pháp thu thập, tổng hợp, trình bày và mô tả các đặc điểm chính của một tập dữ liệu bằng các chỉ số như trung bình, trung vị, độ lệch chuẩn, bảng tần số và biểu đồ.

| | Thống kê mô tả | Thống kê suy luận |
|---|---|---|
| Mục đích | Mô tả dữ liệu hiện có | Suy luận về tổng thể |
| Phạm vi | Tập dữ liệu đang quan sát | Tổng thể dựa trên mẫu |
| Phương pháp | Trung bình, trung vị, biểu đồ | Khoảng tin cậy, kiểm định giả thuyết |
| Ví dụ | Điểm trung bình của 40 sinh viên | Ước lượng điểm trung bình của toàn trường từ mẫu |
#### + Các thước đo thống kê mô tả chính (ví dụ: trung bình, trung vị, phương sai, độ lệch chuẩn) được sử dụng để làm gì? Trong trường hợp nào thì nên dùng trung vị thay vì trung bình?
Mean: Cho biết mức giá trị trung bình của dữ liệu. Tuy nhiên, Mean dễ bị ảnh hưởng bởi các giá trị ngoại lai.

Median: Là giá trị nằm giữa sau khi sắp xếp dữ liệu theo thứ tự tăng dần.

Khi nào nên dùng trung vị thay vì trung bình?
- Khi dữ liệu có ngoại lai lớn.
- Khi phân bố bị lệch trái hoặc lệch phải.
- Khi mô tả thu nhập, giá nhà hoặc thời gian chờ, vì các dữ liệu này thường có một số giá trị rất lớn.

Variance: Đo mức độ phân tán của dữ liệu quanh trung bình, bằng trung bình các bình phương độ lệch. Phương sai càng lớn thì dữ liệu thường càng phân tán. Do bình phương độ lệch được sử dụng, đơn vị của phương sai cũng bị bình phương.

Standard Deviation: Độ lệch chuẩn là căn bậc hai của phương sai. Độ lệch chuẩn có cùng đơn vị với dữ liệu ban đầu nên dễ diễn giải hơn phương sai.
#### + Làm thế nào để xác định phân bố của một tập dữ liệu? Các loại phân bố phổ biến là gì (ví dụ: phân bố chuẩn, lệch trái, lệch phải)?
Có thể xác định phân bố bằng cách:
- Vẽ histogram để quan sát hình dạng dữ liệu.
- Tính trung bình, trung vị, độ lệch chuẩn và các tứ phân vị.
- Quan sát độ lệch, các đỉnh và những giá trị ngoại lai.
- Khi cần kiểm tra một giả định phân phối cụ thể, sử dụng biểu đồ Q–Q hoặc kiểm định thống kê phù hợp.

Các loại phân bố phổ biến:
- Phân bố chuẩn: Có hình dạng gần đối xứng, giống hình chuông. Với phân bố chuẩn lý tưởng, trung bình, trung vị và mode bằng nhau.
- Phân bố lệch trái: Đuôi phân bố kéo dài về bên trái. Một số giá trị thấp kéo trung bình xuống; trong trường hợp lệch trái điển hình, trung bình nhỏ hơn trung vị.
- Phân bố lệch phải: Đuôi phân bố kéo dài về bên phải. Một số giá trị lớn kéo trung bình lên; trong trường hợp lệch phải điển hình, trung bình lớn hơn trung vị.
- Phân bố hai đỉnh: Có hai vùng tập trung giá trị rõ rệt. Điều này có thể gợi ý dữ liệu gồm hai nhóm khác nhau.
#### + Độ lệch chuẩn và phạm vi (range) có ý nghĩa gì trong việc đánh giá sự phân tán của dữ liệu?
Range là hiệu giữa giá trị lớn nhất và nhỏ nhất.

Độ lệch chuẩn đo mức độ các giá trị phân tán quanh trung bình:
- Độ lệch chuẩn nhỏ: dữ liệu thường tập trung gần trung bình.
- Độ lệch chuẩn lớn: dữ liệu phân tán rộng hơn quanh trung bình.
#### + Sự khác biệt giữa các thước đo như Q1, Q2, Q3 trong biểu đồ hộp (boxplot) là gì?
Các tứ phân vị chia dữ liệu đã sắp xếp thành bốn phần, mỗi phần chứa xấp xỉ 25% số quan sát.
- Q1: tứ phân vị thứ nhất. Khoảng 25% dữ liệu nhỏ hơn hoặc bằng Q1.
- Q2: trung vị. Khoảng 50% dữ liệu nhỏ hơn hoặc bằng Q2.
- Q3: tứ phân vị thứ ba. Khoảng 75% dữ liệu nhỏ hơn hoặc bằng Q3.
#### + Làm thế nào để xử lý giá trị thiếu (missing values) trước khi tính toán các chỉ số thống kê mô tả?
| Phương pháp | Khi nào sử dụng? |
|---|---|
| Loại bỏ dòng thiếu | Số dòng thiếu ít và việc loại bỏ không gây sai lệch đáng kể |
| Điền trung bình | Dữ liệu số tương đối đối xứng, ít ngoại lai |
| Điền trung vị | Dữ liệu số lệch hoặc có ngoại lai |
| Điền mode | Dữ liệu phân loại |
| Giữ nguyên và phân tích | Khi bản thân việc thiếu dữ liệu có ý nghĩa |
#### + Bạn có thể giải thích cách đọc và diễn giải một biểu đồ histogram hoặc boxplot từ dữ liệu thực tế không?
Khi đọc histogram:
- Vị trí tập trung: phần lớn dữ liệu nằm trong khoảng nào?
- Hình dạng: đối xứng, lệch trái, lệch phải hay có nhiều đỉnh?
- Độ phân tán: dữ liệu trải rộng hay tập trung?
- Khoảng trống và giá trị cực đoan: có vùng ít dữ liệu hoặc các giá trị bất thường không?

Khi đọc boxplot:
- Hộp càng dài thì IQR càng lớn, tức 50% dữ liệu trung tâm càng phân tán.
- Đường trung vị không nằm giữa hộp có thể gợi ý phân bố không đối xứng.
- Râu dài không đều có thể gợi ý đuôi phân bố khác nhau.
- Các điểm nằm ngoài râu là những quan sát cần kiểm tra thêm, không nhất thiết là dữ liệu sai.
#### + Khi gặp một tập dữ liệu có giá trị ngoại lai (outliers), bạn sẽ xử lý chúng như thế nào trước khi thực hiện thống kê mô tả?
Nếu phát hiện ngoại lai, không nên xóa ngay. Chẳng hạn, một mức lương rất cao có thể là giá trị hợp lệ của quản lý cấp cao. Xóa nó có thể làm sai lệch kết luận về thu nhập của toàn bộ nhân viên.
- Xác minh giá trị có phải do nhập sai, sai đơn vị hoặc là một quan sát hợp lệ.
- Sửa lỗi, giữ nguyên, loại bỏ có căn cứ, biến đổi dữ liệu hoặc giới hạn giá trị cực đoan nếu phù hợp.
- Tính lại các thống kê và kiểm tra xem kết luận có thay đổi đáng kể hay không.

## 1.2. XỬ LÝ VÀ TRỰC QUAN HÓA DỮ LIỆU
### 1.2.1. Ôn tập lý thuyết
#### + Trực quan hóa dữ liệu có vai trò gì trong phân tích dữ liệu? Tại sao nó quan trọng trong khám phá dữ liệu (EDA)?
Trực quan hóa dữ liệu là quá trình biểu diễn dữ liệu dưới dạng hình ảnh như biểu đồ, đồ thị hoặc bản đồ để giúp con người dễ dàng hiểu, phân tích và rút ra kết luận.

Trực quan hóa đặc biệt quan trọng trong khám phá dữ liệu vì giúp:
- Hiểu cấu trúc dữ liệu: nhận biết số lượng mẫu, kiểu dữ liệu và phạm vi giá trị.
- Phát hiện xu hướng: quan sát sự tăng giảm theo thời gian hoặc giữa các nhóm.
- Phát hiện ngoại lai: tìm các giá trị khác biệt so với phần lớn dữ liệu.
- Tìm mối quan hệ: xem xét sự liên hệ giữa các đặc trưng.
- Kiểm tra chất lượng dữ liệu: phát hiện giá trị bất thường, thiếu dữ liệu hoặc phân bố không hợp lý.
- Hỗ trợ lựa chọn mô hình: hiểu đặc điểm dữ liệu để chọn phương pháp phân tích hoặc thuật toán học máy phù hợp.

#### + Các loại biểu đồ phổ biến (như histogram, scatter plot, boxplot, bar chart) được sử dụng trong các trường hợp nào?
- Histogram: Dùng cho dữ liệu số để phát hiện phân bố lệch, nhiều đỉnh hoặc dữ liệu tập trung.
- Scatter plot: Dùng để quan sát mối quan hệ giữa hai biến số.
- Boxplot: Dùng để tóm tắt phân bố dữ liệu số. Hữu ích khi so sánh nhiều nhóm.
- Bar chart: Dùng để so sánh các nhóm hoặc danh mục.
#### + Làm thế nào để chọn loại biểu đồ phù hợp với đặc điểm của dữ liệu (ví dụ: dữ liệu phân loại, dữ liệu số, dữ liệu thời gian)?
| Loại dữ liệu | Mục đích | Biểu đồ phù hợp |
|---|---|---|
| Một biến số | Xem phân bố | Histogram, boxplot |
| Một biến phân loại | Xem tần suất từng nhóm | Bar chart |
| Hai biến số | Xem mối quan hệ | Scatter plot |
| Biến phân loại và biến số | So sánh giữa các nhóm | Boxplot, bar chart |
| Dữ liệu thời gian | Xem xu hướng | Line chart |
| Nhiều biến số | Xem tương quan | Heatmap |
| Tỷ lệ giữa các nhóm | Xem cơ cấu | Stacked bar chart, pie chart khi ít nhóm |
#### + Sự khác biệt giữa các thư viện trực quan hóa trong Python như Matplotlib, Seaborn và Plotly là gì?
| | Matplotlib | Seaborn | Plotly |
|---|---|---|---|
| Độ linh hoạt | Rất cao | Cao | Cao |
| Dễ tạo biểu đồ thống kê | Khá | Rất dễ | Dễ |
| Tương tác | Hạn chế theo cấu hình | Chủ yếu tĩnh | Tốt |
| Xuất PNG/PDF | Có | Có, qua Matplotlib | Có, tùy cấu hình |
| Xuất HTML tương tác | Không phải thế mạnh | Không phải thế mạnh | Có |
#### + Những nguyên tắc thiết kế nào cần tuân thủ để tạo ra một biểu đồ trực quan hóa dễ hiểu và hiệu quả?
Một biểu đồ tốt cần giúp người xem hiểu thông tin nhanh và chính xác:
- Chọn đúng loại biểu đồ: không dùng biểu đồ chỉ vì đẹp nếu nó không trả lời đúng câu hỏi.
- Đặt tiêu đề rõ ràng: thể hiện nội dung và phạm vi dữ liệu.
- Ghi nhãn trục: nêu tên biến, đơn vị đo và khoảng thời gian nếu cần.
- Sử dụng màu sắc hợp lý: hạn chế quá nhiều màu, bảo đảm khả năng phân biệt và thân thiện với người mù màu.
- Giữ tỷ lệ trục trung thực: đặc biệt với biểu đồ cột, tránh cắt trục gây hiểu nhầm về độ lớn chênh lệch.
- Hạn chế yếu tố trang trí thừa: tránh hiệu ứng 3D, lưới quá đậm hoặc quá nhiều nhãn.
- Hiển thị dữ liệu dễ đọc: điều chỉnh kích thước hình, cỡ chữ và chú giải.
- Nêu nguồn và thời điểm dữ liệu: khi biểu đồ dùng trong báo cáo hoặc nghiên cứu.

```python
scores = [5, 6, 7, 8, 6, 9, 7, 8, 5, 10,
          6, 7, 8, 9, 7, 6, 8, 9, 4, 7]

plt.figure(figsize=(8, 5))

plt.hist(scores, bins=6, edgecolor="black")

plt.title("Phan bo diem thi sinh vien")
plt.xlabel("Diem thi")
plt.ylabel("Tan suat")
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

### 1.2.2. Bài làm mẫu
#### Bài toán 1: Thực hiện các nhiệm vụ trong bài toán để làm quen với các công cụ trực quan hóa dữ liệu.
#### Dữ liệu thực hiện là dữ liệu về giá nhà lấy từ https://www.kaggle.com/datasets/thomasnibb/amsterdam-house-price-prediction
Nhiệm vụ 1:

```python
import pandas as pd
houseprices_data = pd.read_csv("HousingPrices-Amsterdam-August-2021.csv")
houseprices_data = houseprices_data[['Zip', 'Price', 'Area', 'Room']]
# Create a PriceperSqm variable based on the Price and Area variables:
houseprices_data['PriceperSqm'] = houseprices_data['Price']/houseprices_data['Area']
```

```python
houseprices_sorted = houseprices_data.sort_values('Price', ascending = False)
houseprices_sorted.head()
# case 1: basic
plt.figure(figsize= (12,6))
x = houseprices_sorted['Zip'][0:10]
y = houseprices_sorted['Price'][0:10]
plt.bar(x,y)
plt.show()
# case 2: advanced 1
plt.figure(figsize= (12,6))
plt.bar(x,y)
plt.title('Top 10 Areas with the highest house prices', fontsize=15)
plt.xlabel('Zip code', fontsize = 12)
plt.xticks(fontsize=10)
plt.ylabel('House prices in millions', fontsize=12)
plt.yticks(fontsize=10)
plt.show()
# case 3: advanced 2
fig, ax = plt.subplots(figsize=(40,18))
x = houseprices_sorted['Zip'][0:10]
y = houseprices_sorted['Price'][0:10]
y1 = houseprices_sorted['PriceperSqm'][0:10]
plt.subplot(1,2,1)
plt.bar(x,y)
plt.xticks(fontsize=17)
plt.ylabel('House prices in millions', fontsize=25)
plt.yticks(fontsize=20)
plt.title('Top 10 Areas with the highest house prices',fontsize=25)
plt.subplot(1,2,2)
plt.bar(x,y1)
plt.xticks(fontsize=17)
plt.ylabel('House prices per sqm', fontsize=25)
plt.yticks(fontsize=20)
plt.title('Top 10 Areas with the highest house prices per sqm',fontsize=25)
plt.show()
```

## 1.3. Phân tích đơn biến và hai biến
### 1.3.1. Ôn lý thuyết
#### + Phân tích đơn biến (univariate analysis) là gì? Nó khác gì với phân tích hai biến (bivariate analysis) trong khám phá dữ liệu?
Trong khám phá dữ liệu (EDA), phân tích đơn biến giúp hiểu đặc điểm của từng biến.
| Tiêu chí | Đơn biến | Hai biến |
|---|---|---|
| Số biến | 1 | 2 |
| Mục tiêu | Hiểu đặc điểm từng biến | Tìm mối quan hệ giữa hai biến |
| Thống kê thường dùng | Mean, median, mode, variance, std | Correlation, covariance, kiểm định thống kê |
| Biểu đồ thường dùng | Histogram, boxplot, bar chart | Scatter plot, heatmap, grouped boxplot |
#### + Các thước đo thống kê nào thường được sử dụng trong phân tích đơn biến (ví dụ: trung bình, trung vị, mode, độ lệch chuẩn)?
| Thước đo | Ý nghĩa |
|---|---|
| Mean | Giá trị trung bình của dữ liệu. |
| Median | Giá trị ở giữa khi sắp xếp dữ liệu. |
| Mode | Giá trị xuất hiện thường xuyên nhất. |
| Variance | Đo mức độ phân tán của dữ liệu quanh trung bình. |
| Standard deviation | Đo mức độ phân tán, có cùng đơn vị với dữ liệu. |
| Minimum / Maximum | Giá trị nhỏ nhất và lớn nhất. |
| Quartiles | Chia dữ liệu đã sắp xếp thành bốn phần. |
| IQR | Khoảng tứ phân vị, bằng Q3 - Q1, thường dùng phát hiện ngoại lệ. |
| Skewness | Cho biết phân phối lệch trái hay lệch phải. |
#### + Trong phân tích hai biến, làm thế nào để xác định mối quan hệ giữa hai biến (ví dụ: tương quan, nhân quả)?
Tương quan cho biết chiều hướng và mức độ liên hệ, thường dùng với hai biến số:
- Tương quan dương: một biến tăng thì biến kia có xu hướng tăng.
- Tương quan âm: một biến tăng thì biến kia có xu hướng giảm.
- Tương quan gần 0: không có bằng chứng về mối liên hệ tuyến tính đáng kể; tuy nhiên, vẫn có thể tồn tại mối quan hệ phi tuyến.

Hai hệ số thường gặp:
- Pearson: đo mức độ liên hệ tuyến tính giữa hai biến số.
- Spearman: đo mức độ liên hệ đơn điệu dựa trên thứ hạng; hữu ích khi dữ liệu không đáp ứng tốt giả định của Pearson hoặc có quan hệ không tuyến tính nhưng đơn điệu.
#### + Sự khác biệt giữa tương quan (correlation) và hiệp biến (covariance) trong phân tích hai biến là gì?
| | Covariance | Correlation |
|---|---|---|
| Ý nghĩa | Hướng thay đổi cùng nhau | Hướng và mức độ liên hệ tuyến tính |
| Giá trị | Không bị giới hạn trong một khoảng cố định | Từ -1 đến 1 |
| Đơn vị | Tích đơn vị của hai biến | Không có đơn vị |
| So sánh giữa các cặp biến | Khó vì phụ thuộc thang đo | Dễ hơn vì đã chuẩn hóa |
#### + Khi nào nên sử dụng biểu đồ trực quan hóa trong phân tích đơn biến so với phân tích hai biến?
- Biểu đồ đơn biến: Dùng khi cần hiểu phân phối của một biến riêng lẻ.
- Biểu đồ hai biến: Dùng khi cần kiểm tra sự liên hệ giữa hai biến.
#### + Đoạn code mẫu để tạo biểu đồ scatter plot hoặc heatmap để phân tích mối quan hệ giữa hai biến?
Mẫu Heatmap:

#### + Làm thế nào để trực quan hóa mối quan hệ giữa một biến số và một biến phân loại bằng biểu đồ boxplot hoặc violin plot trong Python?
- Boxplot giúp so sánh trung vị, khoảng tứ phân vị, độ phân tán và các điểm ngoại lệ của từng nhóm.
- Violin plot thể hiện hình dạng phân phối và mật độ dữ liệu. Phần phình rộng cho biết dữ liệu tập trung nhiều hơn ở vùng giá trị đó. Tham số inner="quartile" hiển thị các tứ phân vị bên trong; cut=0 giới hạn phần mật độ tại phạm vi dữ liệu quan sát.
