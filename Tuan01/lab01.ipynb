## Bài 4. Slicing và Indexing

```python
import torch
a = torch.randn(3, 4, 5)
print("Original tensor:")
print(a)
print("\nTensor a[1]:")
print(a[1])
print("\nTensor a[1:, 2:4]:")
print(a[1:, 2:4])

#shape a[1]
print(a[1].shape)
#shape a[1:, 2:4]
print(a[1:, 2:4].shape)
#lấy 2 phần tử cuối cùng trên chiều thứ 3
print(a[:, :, -2:])
```

## Bài 5. Reshape và Flatten

```python
import torch
a = torch.randn(3, 4)
print("Original:")
print(a)
print("\nFlatten:")
print(a.ravel())
print("\nReshape to 3 x 2 x 2:")
print(a.reshape(3, 2, 2))

#số phần tử trước reshape:
print(a.numel())
#số phần tử sau reshape:
print(a.reshape(3, 2, 2).numel())
#reshape thành 2 x 6:
print(a.reshape(2, 6))
```

## Bài 6. Phép toán và nhân ma trận

```python
import torch
a = torch.randn(2, 3)
b = torch.randn(2, 3)
print("a:")
print(a)
print("b:")
print(b)
print("\na + b:")
print(a + b)
print("\na / b:")
print(a / b)
print("\na^2:")
print(a ** 2)
print("\nMatrix multiplication:")
print(a @ b.T)

#a @ b.T là phép nhân ma trận, a * b là phép nhân từng phần tử
#shape của a @ b.T:
print((a @ b.T).shape)
```

## Bài 7. Các hàm thống kê trên Tensor

```python
import torch
a = torch.randn(3, 4)
print(a)
print("\nMean:")
print(torch.mean(a, dim=0))
print("\nStandard deviation:")
print(torch.std(a, dim=0))
print("\nCumulative sum:")
print(torch.cumsum(a, dim=0))

#thay thành dim=1
print("\nMean:")
print(torch.mean(a, dim=1))
print("\nStandard deviation:")
print(torch.std(a, dim=1))
print("\nCumulative sum:")
print(torch.cumsum(a, dim=1))
#dim=0 là theo cột, dim=1 là theo hàng
```

## Bài 8. Autograd cơ bản

```python
import torch
x = torch.tensor(
3.6,
requires_grad=True
)
y = x * x
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)

#đạo hàm của y = x^2 khi x = 3.6 là 2 * 3.6 = 7.2, đúng với x.grad
#đổi hàm thành y = 3*x^2 + 2x + 1:
y = 3 * x * x + 2 * x + 1
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

## Bài 9. Gradient Descent tìm hệ số đa thức

```python
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
X * X,
X,
np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
y_pred = x @ w
mse = torch.mean(torch.square(y - y_pred))
mse.backward()
optimizer.step()
print("Learned coefficients:")
print(w)

#các hệ số học được gần với hệ số ban đầu nhưng luôn lớn hơn một chút
#vai trò của optimizer.zero_grad() là xóa gradient cũ trước khi tính toán gradient mới, tránh việc cộng dồn gradient từ các bước trước đó.
#vai trò của mse.backward() là tính toán gradient của hàm mất mát với respect to các tham số và lưu trữ chúng trong w.grad.
#vai trò của optimizer.step() là cập nhật các tham số dựa trên gradient đã tính toán và thuật toán tối ưu hóa.
```

## Bài 10. Giải hệ phương trình bằng Autograd

```python
import random
import torch
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()
print("A =", A.item())
print("B =", B.item())
print("C =", C.item())
print("D =", D.item())

#kiểm tra nghiệm: A + B = 9, C - D = 1, A + C = 8, B - D = 2
print("Kiểm tra nghiệm:")
print("A + B =", (A + B).item())
print("C - D =", (C - D).item())
print("A + C =", (A + C).item())
print("B - D =", (B - D).item())
```
